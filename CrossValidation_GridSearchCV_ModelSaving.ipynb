# Week 4: Cross-validation, GridSearchCV, and Model Saving

**Objective:** Evaluate a machine-learning model with cross-validation, tune an SVM using `GridSearchCV`, save the trained model with `joblib`, and load it to predict new data.

**Dataset:** Iris dataset (built into scikit-learn).  
**Libraries:** `scikit-learn`, `joblib`

## 1. Import libraries and load the dataset

````python
from sklearn.datasets import load_iris
from sklearn.model_selection import cross_val_score, GridSearchCV, train_test_split
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score
import joblib

# Load the Iris dataset
iris = load_iris()
X = iris.data
y = iris.target

print("Features:", iris.feature_names)
print("Classes:", iris.target_names)
print("Dataset shape:", X.shape)
````

## 2. Practice Question 1 — Use `cross_val_score` to test accuracy

Cross-validation splits the data into several folds. The model is trained on some folds and tested on the remaining fold, repeating this process so each fold is used for testing.

Here, 5-fold cross-validation is used to estimate the SVM model's accuracy.

````python
svm = SVC(kernel="linear", random_state=42)

scores = cross_val_score(svm, X, y, cv=5, scoring="accuracy")

print("Cross-validation scores:", scores)
print("Mean accuracy: {:.2f}%".format(scores.mean() * 100))
print("Standard deviation: {:.4f}".format(scores.std()))
````

## 3. Practice Question 2 — Implement `GridSearchCV` for SVM

`GridSearchCV` tests the specified combinations of hyperparameters and selects the combination with the best cross-validation score.

````python
# Keep a separate test set for final evaluation
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

parameter_grid = {
    "C": [0.1, 1, 10, 100],
    "kernel": ["linear", "rbf"],
    "gamma": ["scale", "auto"]
}

grid_search = GridSearchCV(
    estimator=SVC(),
    param_grid=parameter_grid,
    cv=5,
    scoring="accuracy",
    n_jobs=-1
)

grid_search.fit(X_train, y_train)

print("Best parameters:", grid_search.best_params_)
print("Best cross-validation accuracy: {:.2f}%".format(
    grid_search.best_score_ * 100
))

best_model = grid_search.best_estimator_
test_predictions = best_model.predict(X_test)
test_accuracy = accuracy_score(y_test, test_predictions)

print("Test accuracy: {:.2f}%".format(test_accuracy * 100))
````

## 4. Practice Question 3 — Save the trained model using `joblib`

The best SVM model is saved to a file named `iris_svm_model.joblib`. Run this cell in the folder where you want the model file stored.

````python
model_path = "iris_svm_model.joblib"
joblib.dump(best_model, model_path)

print("Model saved to:", model_path)
````

## 5. Practice Question 4 — Load the model and predict new data

The new sample must contain four values in the same order as the Iris dataset's features:
1. Sepal length (cm)
2. Sepal width (cm)
3. Petal length (cm)
4. Petal width (cm)

The values below are an example input. Replace them with another sample if needed.

````python
loaded_model = joblib.load("iris_svm_model.joblib")

new_sample = [[5.1, 3.5, 1.4, 0.2]]
predicted_class = loaded_model.predict(new_sample)[0]

print("Predicted class number:", predicted_class)
print("Predicted flower:", iris.target_names[predicted_class])
````

## Result

- `cross_val_score` estimates model accuracy across multiple folds.
- `GridSearchCV` finds the best SVM hyperparameters from the supplied grid.
- `joblib.dump()` saves the fitted model.
- `joblib.load()` reloads the model so it can predict a new sample.
